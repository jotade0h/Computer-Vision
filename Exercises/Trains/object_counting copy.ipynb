```python
import cv2
import numpy as np
from ultralytics import YOLO

# Cargar modelo
model = YOLO("C:/Users/jdoso/Desktop/Yolov8/ultralytics-main/runs/segment/train33/weights/best.pt")  # Cambia la ruta a tu modelo

# Definir zonas prohibidas
#P1, P2, 
zona_1 = np.array([[1090, 100], [1110, 100], [920, 800], [750, 800]], dtype=np.int32)
zona_2 = np.array([[1155, 100], [1175, 100], [1480, 800], [1300, 800]], dtype=np.int32)




# Colores
color_alerta = (0, 0, 255)      # Rojo
color_seguro = (0, 255, 0)      # Verde
color_zona = (0, 0, 255, 80)    # Rojo semitransparente

# Cargar imagen o frame de video

cap = cv2.VideoCapture("C:/Users/jdoso/Desktop/Yolov8/Train/videos/loco_21.mp4")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    overlay = frame.copy()

    # Dibujar zonas prohibidas
    cv2.fillPoly(overlay, [zona_1], color_alerta)
    cv2.fillPoly(overlay, [zona_2], color_alerta)
    frame = cv2.addWeighted(overlay, 0.3, frame, 0.7, 0)

    # Inferencia
    results = model.predict(source=frame, save=False, conf=0.5, verbose=False)[0]

    hay_alerta = False  # Indicador para texto global

    if results.boxes is not None:
        for i, box in enumerate(results.boxes.xyxy.cpu().numpy()):
            x1, y1, x2, y2 = map(int, box[:4])
            cx, cy = int((x1 + x2) / 2), int((y1 + y2) / 2)
            centro = (cx, cy)

            en_zona_1 = cv2.pointPolygonTest(zona_1, centro, False) >= 0
            en_zona_2 = cv2.pointPolygonTest(zona_2, centro, False) >= 0

            if en_zona_1 or en_zona_2:
                hay_alerta = True

            color = color_alerta if (en_zona_1 or en_zona_2) else color_seguro
            texto = "⚠️ Zona prohibida" if (en_zona_1 or en_zona_2) else "OK"

            cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
            cv2.circle(frame, centro, 4, color, -1)
            cv2.putText(frame, texto, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)

    # Dibujar segmentación
    if results.masks is not None:
        for mask in results.masks.data.cpu().numpy():
            mask = cv2.resize(mask, (frame.shape[1], frame.shape[0]))
            mask = (mask > 0.5).astype(np.uint8)
            colored_mask = np.zeros_like(frame)
            colored_mask[:, :] = (255, 0, 255)
            frame = np.where(mask[:, :, None], cv2.addWeighted(frame, 0.5, colored_mask, 0.5, 0), frame)

    # Texto único de alerta si hay cualquier objeto en cualquier zona
    if hay_alerta:
        cv2.putText(frame, "⚠️ Objeto en zona prohibida", (50, frame.shape[0] - 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, color_alerta, 2)

    cv2.imshow("Detección y Segmentación", frame)
    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

# Ultralytics Object Counting

Counting objects using Ultralytics YOLOv8 entails the precise detection and enumeration of specific objects within videos and camera streams. YOLOv8 demonstrates exceptional performance in real-time applications, delivering efficient and accurate object counting across diverse scenarios such as crowd analysis and surveillance. This is attributed to its advanced algorithms and deep learning capabilities.

```python
from ultralytics import YOLO
from ultralytics.solutions import object_counter
import cv2

model = YOLO("yolov8n.pt")
#cap = cv2.VideoCapture('rtsp://admin:william1560@192.168.128.4:554/Streaming/channels/301/')
cap = cv2.VideoCapture("C:/Users/jdoso/Desktop/Yolov8/Train/videos/loco.mp4")
assert cap.isOpened(), "Error reading video file"
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))

# Define region points
region_points = [(20, 400), (1080, 404), (1080, 360), (20, 360)]

# Video writer
video_writer = cv2.VideoWriter("object_counting_output.avi",
                       cv2.VideoWriter_fourcc(*'mp4v'),
                       fps,
                       (w, h))

# Init Object Counter
counter = object_counter.ObjectCounter()
counter.set_args(view_img=True,
                 reg_pts=region_points,
                 classes_names=model.names,
                 draw_tracks=True)

while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        print("Video frame is empty or video processing has been successfully completed.")
        break
    tracks = model.track(im0, persist=True, show=False)

    im0 = counter.start_counting(im0, tracks)
    video_writer.write(im0)

cap.release()
video_writer.release()
cv2.destroyAllWindows()
```
